to test it on a random vector, but your code returns a `MethodError`.

```julia
# function to remove mean from a vector
function remove_mean(vect)
    # fucntion to compute the mean
    function compute_mean(vect)
        element_sum = 0 # initialize sum
        # compute mean and return
        for v in vect
            element_sum += v
        end
        return element_sum / length(vect)
    end

    m = compute_mean(vect) # compute mean
    # return demeaned vector
    return vect - m
end

random_vect = rand(1_000)
@show remove_mean(random_vect)
```

```julia
# function to remove mean from a vector
function remove_mean(vect)
    # fucntion to compute the mean
    function compute_mean(vect)
        element_sum = 0 # initialize sum
        # compute mean and return
        for v in vect
            element_sum += v
        end
        return element_sum / length(vect)
    end

    m = compute_mean(vect) # compute mean
    # return demeaned vector
    return vect .- m
end

random_vect = rand(1_000)
@show remove_mean(random_vect)
```

What does this `MethodError` mean? Why does this code produce one? Fix
the error and solve the problem.

This method error indicates that the vector doesn't correspond to a Float64 as the format "array .- scalar", or dot syntax, should be used for element-wise subtraction. The second return line tries to directly subtract vect - m, but Julia doesn't recognize this syntax and thinks it is a function that doesn't have a defined method. I fixed this problem by changing the line to "vect .- m" in accordance with the format written in the error message.

```julia
# Create one function with two input variables and two output variables
## Input variable 1: Volume of water sent to treatment by land
## Input variable 2: Volume of water sent to chemical treatment
## Output variable 1: Cost
## Output variable 2: Discharge concentration

# Discharge equation: 0.2*X1 + 0.005*X2 + (100-X1-X2) = YUK_concentration
# Cost equation: (X1^2)/20 * X1 + 1.5 * X2 = Treatment_cost

function cost_and_concentration(X1, X2)
    YUK_concentration = 0.2 .* X1 .+ 0.005 .* X2 .+ (100 .- X1 .- X2)
    Treatment_cost = ((X1.*X1)./20) .* X1 .+ 1.5 .* X2
    return(YUK_concentration, Treatment_cost)
end

X1 = [10, 20]
X2 = [10, 20]

output = cost_and_concentration(X1, X2)
YUK_concentration = [out[1] for out in output]
Treatment_cost = [out[2] for out in output]
@show YUK_concentration
@show Treatment_cost

# Sample volumes
Volumes_treated_on_land = (20, 70)
Volumes_treated_at_chem_plant = (80, 30)

#Find new concentrations and costs
output2 = cost_and_concentration(Volumes_treated_on_land, Volumes_treated_at_chem_plant)
YUK_concentrations = [out[1] for out in output2]
Treatment_costs = [out[2] for out in output2]
@show YUK_concentrations
@show Treatment_costs
```

```julia
# Initial function

function cost_and_concentration(X1, X2)
    # Set limits of X1 and X2 being at least 0
    X1 = max.(0, X1)
    X2 = max.(0, X2)

    # Concentration of YUK
    YUK_concentration = 0.2 .* X1 .+ 0.005 .* X2 .+ (100 .- X1 .- X2)

    # Cost of treatment
    Treatment_cost = ((X1.*X1)./20) .* X1 .+ 1.5 .* X2
    return(YUK_concentration, Treatment_cost)
end

X1 = [10, 20, 30, 40, 50]
X2 = [50, 60, 70, 80, 90]

output = cost_and_concentration(X1, X2)
YUK_concentration = [out[1] for out in output]
Treatment_cost = [out[2] for out in output]
@show YUK_concentration
@show Treatment_cost

# Sample volumes
Volumes_treated_on_land = (20, 70)
Volumes_treated_at_chem_plant = (80, 30)

output2 = cost_and_concentration(Volumes_treated_on_land, Volumes_treated_at_chem_plant)
YUK_concentrations = [out[1] for out in output2]
Treatment_costs = [out[2] for out in output2]
@show YUK_concentrations
@show Treatment_costs

# Apply sampling from Dirichlet distribution
using Random, Distributions
Random.seed!(123)

rand(Dirichlet(3, 1), 1000)
sampling = 100*rand(Dirichlet(3,1), 1000)

x = sampling[1:1000]
y = sampling[1001:2000]

YUK_concentration, Treatment_cost = cost_and_concentration(x, y)
```
